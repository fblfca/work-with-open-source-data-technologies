````python
with open (RAW_DIR / 'items.json', encoding='utf-8') as file:
    json_data = json.load(file)

df_json = pd.DataFrame(json_data['items'])
df_json.head()
````

## Достаем данные из XML

````python
with open (RAW_DIR / 'items.xml', encoding='utf-8') as file:
    xml_data = ET.parse(file)
root = xml_data.getroot()

items = root.findall('item')
print(f'Длина xml файла: {len(items)}')

all_data = []

for item in items:
    data = {}

    for field in item:
        data[field.tag] = field.text

    all_data.append(data)

print(f'Пример собранных данных из xml:\n {all_data[:1]}')

df_xml = pd.DataFrame(all_data)
print(df_xml.columns.to_list())
print(df_xml.dtypes)
print(df_xml.shape)
df_xml.head()
````

````python
def parse_updated_at(value):
    '''Так как параметр format="mixed" в pd.to_datetime 
    не справился с распознаванием форматов из файла, пришлось
    реализовывать эту функцию, которая получает на вход строку
    имитирующую дату или таймстамп, проверяет, является ли она null
    или пустой строкой, после чего проверяет, считался ли корректно
    тип данных datetime, тогда сразу преобразовываем в timestamp,
    в ином случае переходим к анализу строки, прогоняем по всем воз-
    можным форматам и преобразовываем строку в timestamp'''
    from datetime import date, datetime
    
    if pd.isna(value) or str(value).strip() == "":
        return pd.NaT

    if isinstance(value, (date, datetime, pd.Timestamp)):
        timestamp = pd.Timestamp(value)

        if timestamp.tzinfo is not None:
            return (
                timestamp
                .tz_convert("Europe/Moscow")
                .tz_localize(None)
            )

        return timestamp

    value = str(value).strip()

    formats = [
        "%Y-%m-%dT%H:%M:%S%z",
        "%Y-%m-%dT%H:%M:%SZ",
        "%Y-%m-%dT%H:%M:%S",
        "%d.%m.%Y %H:%M:%S",
        "%d.%m.%Y %H:%M",
        "%Y-%m-%d",
        "%d.%m.%Y",
        "%Y/%m/%d",
        "%Y.%m.%d",
        "%d/%m/%Y",
    ]

    for date_format in formats:
        try:
            timestamp = pd.to_datetime(value, format=date_format)

            if timestamp.tzinfo is not None:
                timestamp = (
                    timestamp
                    .tz_convert("Europe/Moscow")
                    .tz_localize(None)
                )

            return timestamp

        except (ValueError, TypeError):
            continue

    return pd.NaT

def normalize(df: pd.DataFrame) -> pd.DataFrame:

    df['id'] = pd.to_numeric(
        df['id'], 
        errors='coerce').astype('Int64')
    
    df['name'] = (
        df['name'].astype('string')
        .str.strip().str.replace(r'\s+', ' ', regex=True)
    )

    df['category'] = (
        df['category'].astype('string')
        .str.strip().str.replace(r'\s+', ' ', regex=True)
        .str.capitalize()
    )

    df['value'] = pd.to_numeric(
        df['value'].astype('string')
        .str.strip().str.replace(r'\s+', ' ', regex=True)
        .str.replace(',', '.', regex=False),
        errors='coerce'
    )

    df['updated_at'] = pd.to_datetime(
        df['updated_at'].apply(parse_updated_at),
        format='%m-%d-%Y %H:%M:%S',
        errors='coerce'
    )

    bools = {
        'True': True,
        'Да': True,
        '1': True,
        'Yes': True,
        'False': False,
        'Нет': False,
        '0': False,
        'No': False
    }

    df['active'] = (
        df['active'].astype('string')
        .str.strip().str.capitalize()
    ).map(bools).astype('boolean')

    df = df.drop_duplicates()

    return df
````

````python
df_cleaned = normalize(df)
df_cleaned.head()
````
